```python
from ultralytics import YOLO
model = YOLO("yolov8n.pt")
```

```python
base_options = python.BaseOptions(model_asset_path='pose_landmarker.task')
options = vision.PoseLandmarkerOptions(
    base_options=base_options,
    running_mode=vision.RunningMode.VIDEO,
    output_segmentation_masks=True
)

detector = vision.PoseLandmarker.create_from_options(options)
```

```python
import cv2
import time
```

```python
cap = cv2.VideoCapture(0)
```

```python
# Thresholds
SLOUCH_Y_THRESHOLD = 0.05
SLOUCH_Z_THRESHOLD = 0.10
LOOK_AWAY_X_THRESHOLD = 0.06
HEAD_DOWN_Y_THRESHOLD = 0.06

# YOLO optimisation
FRAME_SKIP = 5
frame_count = 0
phone_detected = False

# State
base_captured = False
base_posture = None

slouch_time = 0
look_away_time = 0
head_down_time = 0

last_time = time.time()
```

```python
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    h, w, _ = frame.shape

    # PHONE DETECTION (YOLO)
    frame_count += 1

    if frame_count % FRAME_SKIP == 0:
        phone_detected = False
        results = model(frame, conf=0.4, verbose=False)

        for r in results:
            for box in r.boxes:
                if int(box.cls[0]) == 67:  # cell phone class
                    phone_detected = True

    # POSE DETECTION (MediaPipe)
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    mp_image = mp.Image(image_format=mp.ImageFormat.SRGB, data=rgb)

    current_time = time.time()
    delta = current_time - last_time
    last_time = current_time

    timestamp = int(current_time * 1000)
    result = detector.detect_for_video(mp_image, timestamp)

    if result.pose_landmarks:
        landmarks = result.pose_landmarks[0]

        nose = landmarks[0]
        left_shoulder = landmarks[11]
        right_shoulder = landmarks[12]

        shoulder_mid_x = (left_shoulder.x + right_shoulder.x) / 2
        shoulder_y = (left_shoulder.y + right_shoulder.y) / 2
        shoulder_z = (left_shoulder.z + right_shoulder.z) / 2

        for i in range(13):
            lm = landmarks[i]
            x = int(lm.x * w)
            y = int(lm.y * h)
            cv2.circle(frame, (x, y), 2, (0, 255, 0), -1)

        # BASE POSTURE CAPTURE
        if not base_captured:
            cv2.putText(frame, "Sit straight and press 'b'",
                        (20, 40), cv2.FONT_HERSHEY_SIMPLEX,
                        0.8, (0, 255, 0), 2)

            if cv2.waitKey(10) & 0xFF == ord('b'):
                base_posture = {
                    "nose_y": nose.y,
                    "nose_z": nose.z,
                    "nose_x": nose.x,
                    "shoulder_y": shoulder_y,
                    "shoulder_z": shoulder_z
                }
                base_captured = True
                slouch_time = look_away_time = head_down_time = 0
                print("Base posture captured")

        # MONITORING MODE
        else:
            # Slouch detection
            slouching = (
                (nose.y - base_posture["nose_y"]) > SLOUCH_Y_THRESHOLD or
                (base_posture["nose_z"] - nose.z) > SLOUCH_Z_THRESHOLD
            )
            slouch_time = slouch_time + delta if slouching else 0

            # Looking away detection
            look_away = abs(nose.x - shoulder_mid_x) > LOOK_AWAY_X_THRESHOLD
            look_away_time = look_away_time + delta if look_away else 0

            # Head down detection (neutral signal)
            head_down = (nose.y - base_posture["nose_y"]) > HEAD_DOWN_Y_THRESHOLD
            head_down_time = head_down_time + delta if head_down else 0

            # Combine signals to infer phone usage
            likely_phone_usage = (
                phone_detected and
                head_down_time > 3 and
                (look_away_time > 3 or slouch_time > 3)
            )

            # DISPLAY ALERTS
            alert_y = 40
            any_alert = False

            if slouch_time > 1:
                cv2.putText(frame, f"SLOUCHING: {int(slouch_time)}s",
                            (20, alert_y), cv2.FONT_HERSHEY_SIMPLEX,
                            0.9, (0, 0, 255), 2)
                alert_y += 35
                any_alert = True

            if look_away_time > 3:
                cv2.putText(frame, f"LOOKING AWAY: {int(look_away_time)}s",
                            (20, alert_y), cv2.FONT_HERSHEY_SIMPLEX,
                            0.9, (0, 255, 255), 2)
                alert_y += 35
                any_alert = True

            if head_down_time > 3:
                cv2.putText(frame, f"HEAD DOWN: {int(head_down_time)}s",
                            (20, alert_y), cv2.FONT_HERSHEY_SIMPLEX,
                            0.9, (200, 200, 0), 2)
                alert_y += 35
                any_alert = True

            if phone_detected:
                cv2.putText(frame, "PHONE DETECTED",
                            (20, alert_y), cv2.FONT_HERSHEY_SIMPLEX,
                            0.9, (255, 165, 0), 2)
                alert_y += 35
                any_alert = True

            if likely_phone_usage:
                cv2.putText(frame, "LIKELY PHONE USAGE",
                            (20, alert_y), cv2.FONT_HERSHEY_SIMPLEX,
                            0.9, (255, 0, 255), 2)
                any_alert = True

            if not any_alert:
                cv2.putText(frame, "POSTURE OK",
                            (20, 40), cv2.FONT_HERSHEY_SIMPLEX,
                            0.9, (0, 255, 0), 2)

    cv2.imshow("Posture Monitor", frame)

    if cv2.waitKey(10) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
